## Bước 7. Chạy HVI-CIDNet trên các bộ test và chép ảnh ra

Dùng chế độ `--unpaired --custome --custome_path <thư mục ảnh>` của repo (README, mục Testing) để chạy trên bất kỳ thư mục ảnh nào, kể cả ảnh `input` của LOL. Mỗi bộ dùng một file trọng số:

| Bộ | Trọng số |
|---|---|
| LOLv1 | `LOLv1/wo_perc.pth` |
| LOLv2_Real | `LOLv2_real/best_SSIM.pth` |
| DICM, LIME, custom_night | `LOLv2_syn/w_perc.pth` (ví dụ trong README) |

Hàm `chay_va_chep` chạy mô hình, tự tìm thư mục ảnh mới sinh trong `HVI-CIDNet/output/`, chép sang `results/<BỘ>/HVI-CIDNet/` và cảnh báo nếu tên file không khớp ảnh input. **Hãy ghi lại trọng số bạn đã dùng** để đưa vào báo cáo.

Chưa kiểm chứng: chế độ `--custome` có cho ra đúng số như lệnh `--lol` của README hay không (mặc định alpha = 1.0, gamma = 1.0). Nếu số LOLv1 lệch nhiều so với README (khoảng 23.5 không GT Mean, 28.1 có GT Mean), hãy chạy thêm `python eval.py --lol` trong thư mục repo với dữ liệu đặt đúng cấu trúc README để đối chiếu.

````python
import shutil, glob, time, subprocess
IMG = (".png", ".jpg", ".jpeg", ".bmp")

def chay_repo(args, cwd="HVI-CIDNet"):
    """Chạy eval.py CỦA REPO (không phải eval.py của bạn). Trả về các thư mục có ảnh mới sinh ra."""
    t0 = time.time()
    r = subprocess.run([sys.executable, "eval.py"] + args, cwd=cwd)
    if r.returncode != 0:
        print("!! Lệnh của repo báo lỗi (xem thông báo ở trên). Dán lỗi này cho Claude.")
        return []
    moi = [f for f in glob.glob(os.path.join(cwd, "output", "**", "*"), recursive=True)
           if f.lower().endswith(IMG) and os.path.getmtime(f) >= t0 - 1]
    return sorted({os.path.dirname(f) for f in moi})

def chep_ket_qua(thu_muc_anh_ra, bo, ten_pp="HVI-CIDNet"):
    dich = os.path.join("results", bo, ten_pp)
    os.makedirs(dich, exist_ok=True)
    n = 0
    for f in glob.glob(os.path.join(thu_muc_anh_ra, "*")):
        if f.lower().endswith(IMG):
            shutil.copy(f, dich); n += 1
    print(f"{bo}: chép {n} ảnh -> {dich}")
    # kiểm tra tên file khớp ảnh input
    inp = {os.path.splitext(f)[0] for f in os.listdir(os.path.join("test_cases", bo, "input"))}
    out = {os.path.splitext(f)[0] for f in os.listdir(dich)}
    if inp - out:
        print("  !! Thiếu ảnh ra cho:", sorted(inp - out)[:5], "... (đổi tên file ảnh ra cho trùng ảnh input)")
    return dich

def chay_va_chep(bo, trong_so, them=()):
    dau_vao = os.path.abspath(os.path.join("test_cases", bo, "input"))
    thu_muc = chay_repo(["--unpaired", "--custome", "--custome_path", dau_vao,
                         "--unpaired_weights", trong_so, *them])
    print("Ảnh mới nằm ở:", thu_muc)
    if len(thu_muc) == 1:
        chep_ket_qua(thu_muc[0], bo)
    else:
        print("Không xác định được 1 thư mục duy nhất. Mở HVI-CIDNet/output, rồi gọi chep_ket_qua('<thư mục>', '%s')." % bo)
````

````python
# Chạy từng bộ (trọng số tính từ thư mục HVI-CIDNet). Bộ nào chưa có ảnh input thì bỏ dấu # ở bộ đã sẵn sàng.
chay_va_chep("LOLv1",        "./weights/LOLv1/wo_perc.pth")
# chay_va_chep("LOLv2_Real", "./weights/LOLv2_real/best_SSIM.pth")
# chay_va_chep("DICM",       "./weights/LOLv2_syn/w_perc.pth")
# chay_va_chep("LIME",       "./weights/LOLv2_syn/w_perc.pth")
# chay_va_chep("custom_night", "./weights/LOLv2_syn/w_perc.pth")
````

Nếu ô trên báo lỗi **thiếu file trọng số** hoặc **sai tên tham số**: mở `HVI-CIDNet/eval.py` xem repo nạp đường dẫn nào và nhận tham số gì, rồi dán lỗi cho Claude. Ảnh CLAHE/Gamma tự code của TV1 được tạo ở Bước 8, bạn không cần chép tay.

**Ca 2 (độ sáng/α), phần B:** chạy mô hình với nhiều giá trị `--alpha` (thang độ sáng của repo) trên ảnh của ca 2, mỗi giá trị một thư mục kết quả `HVI-CIDNet_a<alpha>`. Ảnh nên đặt ở `test_cases/failure_cases/case2_k_alpha/input/`. (README: `--alpha` là hệ số độ sáng; riêng `alpha_s` chỉ có trong `eval_hf.py`.)

````python
CA2 = "case2_k_alpha"
dau_vao = os.path.abspath(os.path.join("test_cases", "failure_cases", CA2, "input"))
for a in ["0.5", "1.0", "1.5"]:
    thu_muc = chay_repo(["--unpaired", "--custome", "--custome_path", dau_vao,
                         "--unpaired_weights", "./weights/LOLv2_syn/w_perc.pth", "--alpha", a])
    if len(thu_muc) == 1:
        dich = os.path.join("results", "failure_cases", CA2, "HVI-CIDNet_a" + a)
        os.makedirs(dich, exist_ok=True)
        for f in glob.glob(os.path.join(thu_muc[0], "*")):
            if f.lower().endswith(IMG): shutil.copy(f, dich)
        print("alpha", a, "->", dich)
    else:
        print("alpha", a, ": không xác định được thư mục ảnh ra:", thu_muc)
# Sau đó: PHUONG_PHAP = "opencv_clahe HVI-CIDNet_a0.5 HVI-CIDNet_a1.0 HVI-CIDNet_a1.5" ở ô panel/metrics.
````
